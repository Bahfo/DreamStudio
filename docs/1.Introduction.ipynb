## 1.1 Getting Familiar
DreamStudio isn't very hard to get familiar with. Specfically, once you open the IDE for the first time, you are presented with the following: 
- Left and Right vertical toolboxes on each side of the app.
- Left and Right panels that open and close according to the accompanying button in the toolbox.
- Main Window, which houses the code editor and other utilities.
- Horizontal Panel which includes the terminal, Outputs Widget, Python shell, and more.
- The Title Bar, and Options Bar at the top, and the Status Bar at the bottom.

The title bar contains the main drop-down menu that houses almost all functionalities of the IDE, notice that it has multiple menus specified for each topic: File operations, View options, Code Editing, etc. It also has a menu dedicated for plugins and extensions, and a help menu.

The options bar contains some quick-to-access options that are helpful right away, such as some file and project operations, cut, copy, paste, in addition to run and debug options, and changing themes.

Left and Right vertical sidebars contain buttons that are responsible for opening a corresponding panel next to them. For example, you may encounter The `Explorer` which opens up the *Solution Explorer* widget at the left. Or `Properties` button at the right which opens the *Properties Explorer* widget at the right.

The panels next to each toolbox are what house most of the IDE core functionalities, including *Solution Explorer*, *Server Explorer*, *Git Source Control*, and more.

Main Window is where the code editor lives. Double-click any file from the Solution Explorer, or open a file using Ctrl+O (Opening an Existing File) or Ctrl+N (New File) and it will show up.

Horziontal Panel is where other supportive functionalities live. Including the *Terminal*, *PromptX*, *Outputs Window*, and *Processes Monitor*. 

Finally, the status bar includes other useful-to-look-at options, including a quick-to-switch button to switch branches (if working remotly or locally), problems and warnings, current line and column in an opened editor, and more.
