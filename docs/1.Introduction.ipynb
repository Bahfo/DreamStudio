## Structure of the IDE
DreamStudio is structured to suite programmers with similar UI architecture to already known IDEs in the public. You may ask yourself: What is different within DreamStudio? And the answer is: *Not only DreamStudio gives flexability in usage, not only it simplifies integration in Python environment. It also follows a design philosophy of 'Must be simple enough so a child can learn, but must be professional enough for an expert to use.'*. DreamStudio introduces everything in clear structured parts. Solution explorers, source control, properties, the code editor. Featured with a clean set of keybindings that is easy to change. It also comes configured with a set of well-known keybindings (like Vim KB or VSCode KB). Not only that, but DreamStudio has enough flexibility and versitality to introduce extensions that are downloadable and configurable.

In conclusion, we try to feature DreamStudio as a simple but professional IDE for all peoples tasks.
