# Contributing to DreamStudio


###### This section is about contributing, for the `real` documentation, go to `getting_started`.

Hi and welcome! This tutorial is all about the conditions and requirements needed for any person to contribute in the open source program of DreamStudio. We are open to all contributors! anyone who can meet the requirements is welcomed to join.

First of all, any contributor should be able to understand some basic fundamentals about how code is written:

1. Documentation values the most: A good documentation is required for others to understand your code, we are not asking to be a professional in writing documentations, but at least give a sense of understanding to what you write and what code you develop.

2. Comments are important, but not overwhelming: You should write comments when necessary, you do not need to comment every step you make, but the critical parts where you feel your code is in its climax, that is where comments are necessary. 

3. Do not expect perfections: This project is not a code-perfection. Me -myself- is not a programming-language master, and we make bad decisions here and there. If you are one of the extremes that like to optimize every piece of code, then we will not have a good time. 

4. Optimization at the critical parts only: Only the backbone critical parts are needed to optimize. For example, the code analyzer, regular expressions engine, etc. These are something meaningful to optimize, but not -for example- the menubar UI class. That is something that works fine as it is, and no one complains about its job, so it is fine if you leave it as it is.
